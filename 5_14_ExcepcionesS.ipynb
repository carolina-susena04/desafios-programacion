# Excepciones y errores

## Excepciones

"El agua hierve a 100 grados bajo cero cuando hace calor."

Aunque esta oración es correcta desde el punto de vista gramatical, su significado está lleno de contradicciones. Algo similar ocurre en la programación: incluso cuando el código tiene la sintaxis correcta, puede generar errores lógicos durante la ejecución. A estos errores se les conoce como excepciones, y son detectados cuando el programa se está ejecutando, generando mensajes de error que pueden, en muchos casos, ser controlados y gestionados.

Ejecuta las siguientes líneas y observa que ocurre...

```python
10 * (1/0)
```

```python
4 + spam*3
```

```python
'2' + 2
```

La última línea del mensaje de error es crucial, ya que indica qué tipo de excepción ha ocurrido. Cada tipo de excepción tiene un propósito específico y se utiliza para señalar un tipo particular de error.

# Como corregir esas excepciones

```python
a = (int) (input("Ingrese el dividendo: "))
b = (int) (input("Ingrese el divisor: "))

print(a/b)
```

Esto no funciona porque el usuario puede ingresar en b el valor 0.
Mejoremos el codigo.

```python
# Solicitamos al usuario el dividendo y lo convertimos a entero
a = float(input("Ingrese el dividendo: "))

# Solicitamos al usuario el divisor y lo convertimos a entero
b = float(input("Ingrese el divisor: "))

# Intentamos realizar la división
try:
    resultado = a/b
    print(resultado)
except ZeroDivisionError:  # Si ocurre una división por cero
    print("Error: División por cero no permitida.")
```

```python

# ==========================================
# DESAFÍO 67: MANEJO DE EXCEPCIONES
# DIVISION ENTRE DOS NUMEROS
# ==========================================

print("DESAFÍO 67")

try:
    num1 = int(input("Ingrese el primer número entero: "))
    num2 = int(input("Ingrese el segundo número entero: "))

    resultado = num1 / num2

    print("Resultado de la división:", resultado)

except ValueError:
    print("Error: Debe ingresar números enteros.")

except ZeroDivisionError:
    print("Error: No se puede dividir por cero.")


# ==========================================
# DESAFÍO 68: MANEJO DE VARIAS EXCEPCIONES
# ==========================================

print("\nDESAFÍO 68")

valores = [10, 5, 0, "hola", 20]

for valor in valores:
    try:
        numero = int(valor)
        resultado = 100 / numero
        print("Resultado:", resultado)

    except ZeroDivisionError:
        print("Error: División por cero.")

    except TypeError:
        print("Error: Tipo de dato incorrecto.")

    except ValueError:
        print("Error: Valor no válido.")


# ==========================================
# DESAFÍO 69: CALCULAR EL FACTORIAL
# ==========================================

print("\nDESAFÍO 69")

def factorial(numero):

    if not isinstance(numero, int) or isinstance(numero, bool):
        raise TypeError("El número debe ser entero.")

    if numero < 0:
        raise ValueError("El número no puede ser negativo.")

    if numero > 1000:
        raise OverflowError("El número es demasiado grande.")

    if numero == 0 or numero == 1:
        return 1

    return numero * factorial(numero - 1)


try:
    numero = int(input("Ingrese un número para calcular su factorial: "))

    resultado = factorial(numero)

    print("El factorial es:", resultado)

except ValueError as error:
    print("Error:", error)

except TypeError as error:
    print("Error:", error)

except OverflowError as error:
    print("Error:", error)


# ==========================================
# DESAFÍO 70: EXCEPCIÓN PERSONALIZADA
# NUMEROS NEGATIVOS
# ==========================================

print("\nDESAFÍO 70")

import math

class NegativeNumberError(Exception):
    pass


def calcular_raiz(numero):

    if numero < 0:
        raise NegativeNumberError(
            "No se puede calcular la raíz de un número negativo."
        )

    return math.sqrt(numero)


try:
    numero = float(input("Ingrese un número para calcular su raíz: "))

    resultado = calcular_raiz(numero)

    print("La raíz cuadrada es:", resultado)

except NegativeNumberError as error:
    print("Error:", error)

except ValueError:
    print("Error: Debe ingresar un número válido.")


# ==========================================
# DESAFÍO 71: ABRIR Y LEER UN ARCHIVO
# MANEJO DE EXCEPCIONES Y FINALLY
# ==========================================

print("\nDESAFÍO 71")

archivo = None

try:
    nombre_archivo = input("Ingrese el nombre del archivo: ")

    archivo = open(nombre_archivo, "r", encoding="utf-8")

    contenido = archivo.read()

    print("Contenido del archivo:")
    print(contenido)

except FileNotFoundError:
    print("Error: El archivo no existe.")

except PermissionError:
    print("Error: No tiene permiso para abrir el archivo.")

finally:
    if archivo is not None:
        archivo.close()
        print("El archivo se cerró correctamente.")
    else:
        print("No se pudo abrir el archivo.")
```
