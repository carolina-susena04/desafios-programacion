# Introducción a la Herencia

En nuestro viaje por el mundo de la Programación Orientada a Objetos (POO), hemos creado clases que representan entidades en una biblioteca, como `Autor` y `Libro`. Pero, ¿qué pasa si queremos crear nuevas clases que compartan características con las ya existentes, pero con algunas diferencias o adicionales? ¡Aquí es donde entra en juego la herencia!

### Pregunta problema: 
¿Cómo podemos extender nuestras clases existentes para representar diferentes tipos de autores o libros, sin tener que reescribir todo el código?



```python
# Definimos la clase base Autor
class Autor:
    def __init__(self, nombre):
        self.nombre = nombre
```

## Conceptos clave de la herencia

La **herencia** permite que una clase (llamada subclase o clase hija) herede atributos y métodos de otra clase (llamada superclase o clase base). La herencia facilita la reutilización y extensión del código.

- **Herencia simple:** Una clase deriva de una única clase base.
- **Herencia múltiple:** Una clase puede derivar de varias clases base.
- **super():** Función que permite llamar a métodos de la clase base desde la clase derivada.

```python
# Creamos una subclase Escritor que hereda de Autor
class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero
```

## Ejemplos prácticos de Herencia

Vamos a extender nuestra clase `Autor` para crear una nueva clase `Escritor` que tenga un atributo adicional: el género literario principal en el que escribe.


```python
# Instanciamos un objeto de la clase Escritor
escritor = Escritor("Mario Benedetti", "Realismo Social")
print(escritor.nombre, escritor.genero)
```

## Herencia Múltiple

En Python, una clase puede heredar de varias clases base, lo que se conoce como herencia múltiple. Aunque puede ser útil, también puede llevar a complicaciones si varias clases base tienen atributos o métodos con el mismo nombre.


```python
# Definimos una segunda clase base
class Academico:
    def __init__(self, universidad):
        self.universidad = universidad

# Creamos una clase que hereda de Escritor y Academico
class EscritorAcademico(Escritor, Academico):
    def __init__(self, nombre, genero, universidad):
        Escritor.__init__(self, nombre, genero)
        Academico.__init__(self, universidad)
```

### Desafío 61:
Crea una jerarquía de clases para representar diferentes tipos de empleados en una biblioteca, utilizando herencia múltiple y composición. Por ejemplo, implementa clases como Empleado, Gerente, Tecnico, y Voluntario, donde Gerente y Tecnico hereden de Empleado, y algunos puedan tener roles adicionales mediante composición con otras clases como Administrador o Mantenimiento.



```python

# ==========================================
# DESAFÍO 57: CLASE POETA QUE HEREDA DE AUTOR
# ==========================================

class Autor:

    def __init__(self, nombre, nacionalidad):
        self.nombre = nombre
        self.nacionalidad = nacionalidad
        self.libros = []

    def agregar_libro(self, libro):
        self.libros.append(libro)


class Poeta(Autor):

    def __init__(self, nombre, nacionalidad, tipo_poesia):
        super().__init__(nombre, nacionalidad)
        self.tipo_poesia = tipo_poesia

    def mostrar_datos(self):
        print("Nombre:", self.nombre)
        print("Nacionalidad:", self.nacionalidad)
        print("Tipo de poesía:", self.tipo_poesia)


print("DESAFÍO 57")

poeta = Poeta("Mario Benedetti", "Uruguayo", "Poesía romántica")
poeta.mostrar_datos()


# ==========================================
# DESAFÍO 58: CLASE BIBLIOTECARIO QUE HEREDA
# DE USUARIO
# ==========================================

class Usuario:

    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad


class Bibliotecario(Usuario):

    def __init__(self, nombre, edad, seccion, anios_experiencia):
        super().__init__(nombre, edad)
        self.seccion = seccion
        self.anios_experiencia = anios_experiencia

    def mostrar_datos(self):
        print("Nombre:", self.nombre)
        print("Edad:", self.edad)
        print("Sección:", self.seccion)
        print("Años de experiencia:", self.anios_experiencia)


print("\nDESAFÍO 58")

bibliotecario = Bibliotecario("Carlos", 35, "Literatura", 8)
bibliotecario.mostrar_datos()


# ==========================================
# DESAFÍO 59: LIBRO DIGITAL Y EBOOK
# ==========================================

class Libro:

    def __init__(self, titulo, genero, isbn):
        self.titulo = titulo
        self.genero = genero
        self.isbn = isbn

    def mostrar_informacion(self):
        print("Título:", self.titulo)
        print("Género:", self.genero)
        print("ISBN:", self.isbn)


class LibroDigital(Libro):

    def __init__(self, titulo, genero, isbn, formato, tamano_archivo):
        super().__init__(titulo, genero, isbn)
        self.formato = formato
        self.tamano_archivo = tamano_archivo


class Ebook(LibroDigital):

    def __init__(self, titulo, genero, isbn, formato,
                 tamano_archivo, enlace_descarga):
        super().__init__(titulo, genero, isbn, formato, tamano_archivo)
        self.enlace_descarga = enlace_descarga

    def mostrar_informacion(self):
        super().mostrar_informacion()
        print("Formato:", self.formato)
        print("Tamaño:", self.tamano_archivo, "MB")
        print("Enlace de descarga:", self.enlace_descarga)


print("\nDESAFÍO 59")

ebook = Ebook(
    "Aprendiendo Python",
    "Programación",
    "123456",
    "PDF",
    15,
    "www.ejemplo.com/libro"
)

ebook.mostrar_informacion()


# ==========================================
# DESAFÍO 60: HERENCIA MÚLTIPLE
# ESCRITOR ACADÉMICO
# ==========================================

class Persona:

    def __init__(self, nombre):
        self.nombre = nombre


class Escritor(Persona):

    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero


class Academico(Persona):

    def __init__(self, nombre, universidad):
        super().__init__(nombre)
        self.universidad = universidad


class EscritorAcademico(Escritor, Academico):

    def __init__(self, nombre, genero, universidad):
        super().__init__(nombre, genero)
        self.universidad = universidad
        self.articulos = []

    def publicar_articulo(self, articulo):
        self.articulos.append(articulo)
        print("Artículo publicado:", articulo)

    def mostrar_datos(self):
        print("Nombre:", self.nombre)
        print("Género:", self.genero)
        print("Universidad:", self.universidad)
        print("Artículos:", self.articulos)


print("\nDESAFÍO 60")

escritor = EscritorAcademico(
    "Juan Pérez",
    "Investigación",
    "Universidad Nacional"
)

escritor.publicar_articulo("La programación en Python")
escritor.mostrar_datos()


# ==========================================
# DESAFÍO 61: JERARQUÍA DE EMPLEADOS
# HERENCIA MÚLTIPLE Y COMPOSICIÓN
# ==========================================

class Empleado:

    def __init__(self, nombre, sueldo):
        self.nombre = nombre
        self.sueldo = sueldo

    def mostrar_datos(self):
        print("Nombre:", self.nombre)
        print("Sueldo:", self.sueldo)


class Gerente(Empleado):

    def __init__(self, nombre, sueldo, departamento):
        super().__init__(nombre, sueldo)
        self.departamento = departamento

    def mostrar_datos(self):
        super().mostrar_datos()
        print("Departamento:", self.departamento)


class Tecnico(Empleado):

    def __init__(self, nombre, sueldo, especialidad):
        super().__init__(nombre, sueldo)
        self.especialidad = especialidad

    def mostrar_datos(self):
        super().mostrar_datos()
        print("Especialidad:", self.especialidad)


class Voluntario(Empleado):

    def __init__(self, nombre, sueldo, horas):
        super().__init__(nombre, sueldo)
        self.horas = horas

    def mostrar_datos(self):
        super().mostrar_datos()
        print("Horas de voluntariado:", self.horas)


# Clases para composición

class Administrador:

    def administrar(self):
        print("Realizando tareas administrativas")


class Mantenimiento:

    def reparar(self):
        print("Realizando tareas de mantenimiento")


# Clase que utiliza composición

class EmpleadoConRoles(Empleado):

    def __init__(self, nombre, sueldo):
        super().__init__(nombre, sueldo)
        self.roles = []

    def agregar_rol(self, rol):
        self.roles.append(rol)

    def realizar_tareas(self):
        for rol in self.roles:
            if isinstance(rol, Administrador):
                rol.administrar()

            elif isinstance(rol, Mantenimiento):
                rol.reparar()


print("\nDESAFÍO 61")

gerente = Gerente("Roberto", 50000, "Biblioteca")
gerente.mostrar_datos()

tecnico = Tecnico("Pedro", 30000, "Informática")
tecnico.mostrar_datos()

voluntario = Voluntario("Ana", 0, 20)
voluntario.mostrar_datos()

empleado = EmpleadoConRoles("Luis", 35000)

empleado.agregar_rol(Administrador())
empleado.agregar_rol(Mantenimiento())

print("\nRoles adicionales del empleado:")
empleado.realizar_tareas()
```
