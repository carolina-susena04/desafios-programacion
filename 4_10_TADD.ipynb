# Introducción a los Tipos Abstractos de Datos (TAD)

## Concepto de TAD

Un Tipo Abstracto de Datos (TAD) es una especificación de un conjunto de datos y las operaciones que se pueden realizar con esos datos. El TAD se centra en qué operaciones existen y no en cómo se implementan. Es una herramienta conceptual que nos permite abstraernos de los detalles y pensar en términos de operaciones y datos.

### Ejemplo de TAD: Lista


```python
lista = []

def agregar(elemento):
    lista.append(elemento)

def eliminar(elemento):
    if elemento in lista:
        lista.remove(elemento)

def buscar(elemento):
    return elemento in lista
```

## Desventajas de este enfoque

1. **Falta de encapsulamiento**: En el ejemplo proporcionado, tanto las operaciones como los datos están expuestos y pueden ser modificados directamente desde fuera del TAD. Esto viola el principio de encapsulamiento, que es fundamental para el diseño de software robusto y mantenible.

2. **Dificultad en reutilización**: Con el diseño actual, si quisieras tener múltiples listas y aplicar operaciones sobre ellas, tendrías que crear funciones separadas o modificar las existentes, lo que no es práctico.

3. **Problemas de extensibilidad**: Cambiar la implementación interna (por ejemplo, cambiar de una lista a un árbol binario para mejorar la eficiencia) requeriría modificar todas las funciones que interactúan con la estructura de datos. Esto hace que el código sea difícil de mantener y de extender con nuevas funcionalidades.

4. **Falta de abstracción**: Aunque el TAD debería ayudarnos a abstraernos de los detalles de implementación, el enfoque actual no lo hace de manera efectiva. Esto podría llevar a errores si no se comprenden completamente las implementaciones subyacentes.

### Desafío 45: Cambio frecuente en requisitos
Supón que estás desarrollando un juego de video con distintos tipos de personajes y armas. Los requerimientos cambian con frecuencia, añadiendo nuevos personajes y habilidades. Mantener y actualizar TADs en este escenario podría ser una tarea titánica.

### Desafío 46: Estructuras de datos anidadas
Considera un sistema de manejo de inventario para una cadena de tiendas minoristas. Tienes que tratar con datos de productos, tiendas, empleados, y transacciones, donde cada tienda podría tener múltiples productos y empleados. Gestionar estas relaciones con TADs podría ser ineficiente y propenso a errores.


```python

# ==========================================
# DESAFÍO 44: SISTEMAS CON MÚLTIPLES ENTIDADES
# INTERCONECTADAS
# ==========================================

class Biblioteca:

    def __init__(self):
        self.libros = {}
        self.usuarios = {}
        self.prestamos = []

    def agregar_libro(self, titulo):
        self.libros[titulo] = True

    def agregar_usuario(self, nombre):
        self.usuarios[nombre] = []

    def prestar_libro(self, titulo, usuario):
        if titulo in self.libros and self.libros[titulo]:
            if usuario in self.usuarios:
                self.libros[titulo] = False
                self.usuarios[usuario].append(titulo)
                self.prestamos.append({
                    "libro": titulo,
                    "usuario": usuario
                })
                print("Libro prestado correctamente")
            else:
                print("Usuario no encontrado")
        else:
            print("Libro no disponible")

    def devolver_libro(self, titulo, usuario):
        if usuario in self.usuarios:
            if titulo in self.usuarios[usuario]:
                self.usuarios[usuario].remove(titulo)
                self.libros[titulo] = True
                print("Libro devuelto correctamente")
            else:
                print("El usuario no tiene ese libro")

    def mostrar_libros(self):
        print("Libros:", self.libros)


print("DESAFÍO 44")

biblioteca = Biblioteca()

biblioteca.agregar_libro("Python")
biblioteca.agregar_libro("Java")
biblioteca.agregar_usuario("Juan")

biblioteca.prestar_libro("Python", "Juan")
biblioteca.devolver_libro("Python", "Juan")
biblioteca.mostrar_libros()


# ==========================================
# DESAFÍO 45: CAMBIO FRECUENTE EN REQUISITOS
# ==========================================

class Personaje:

    def __init__(self, nombre, vida):
        self.nombre = nombre
        self.vida = vida
        self.habilidades = []
        self.armas = []

    def agregar_habilidad(self, habilidad):
        self.habilidades.append(habilidad)

    def agregar_arma(self, arma):
        self.armas.append(arma)

    def mostrar_personaje(self):
        print("Personaje:", self.nombre)
        print("Vida:", self.vida)
        print("Habilidades:", self.habilidades)
        print("Armas:", self.armas)


print("\nDESAFÍO 45")

personaje = Personaje("Guerrero", 100)

personaje.agregar_habilidad("Fuerza")
personaje.agregar_habilidad("Velocidad")
personaje.agregar_arma("Espada")
personaje.agregar_arma("Escudo")

personaje.mostrar_personaje()


# ==========================================
# DESAFÍO 46: ESTRUCTURAS DE DATOS ANIDADAS
# ==========================================

class Inventario:

    def __init__(self):
        self.tiendas = {}

    def agregar_tienda(self, nombre):
        self.tiendas[nombre] = {
            "productos": {},
            "empleados": [],
            "transacciones": []
        }

    def agregar_producto(self, tienda, producto, cantidad):
        if tienda in self.tiendas:
            self.tiendas[tienda]["productos"][producto] = cantidad

    def agregar_empleado(self, tienda, empleado):
        if tienda in self.tiendas:
            self.tiendas[tienda]["empleados"].append(empleado)

    def vender_producto(self, tienda, producto, cantidad):
        if tienda in self.tiendas:
            productos = self.tiendas[tienda]["productos"]

            if producto in productos and productos[producto] >= cantidad:
                productos[producto] -= cantidad

                self.tiendas[tienda]["transacciones"].append({
                    "producto": producto,
                    "cantidad": cantidad
                })

                print("Venta realizada correctamente")
            else:
                print("No hay suficiente stock")

    def mostrar_inventario(self, tienda):
        if tienda in self.tiendas:
            print("Inventario de", tienda)
            print(self.tiendas[tienda])


print("\nDESAFÍO 46")

inventario = Inventario()

inventario.agregar_tienda("Tienda Central")

inventario.agregar_producto("Tienda Central", "Computadora", 10)
inventario.agregar_producto("Tienda Central", "Teclado", 20)

inventario.agregar_empleado("Tienda Central", "Carlos")
inventario.agregar_empleado("Tienda Central", "Maria")

inventario.vender_producto("Tienda Central", "Computadora", 2)

inventario.mostrar_inventario("Tienda Central")
```
