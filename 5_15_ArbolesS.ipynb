# Árboles en Python

Los árboles son una estructura de datos fundamental en ciencias de la computación y programación. Un árbol consiste en un conjunto de nodos conectados por bordes, con un nodo raíz desde el cual se derivan los demás nodos. Cada nodo puede tener cero o más nodos hijos.

En esta guía, exploraremos cómo implementar y trabajar con árboles en Python.

## Definición de un Nodo

Primero, definiremos la clase `Nodo` que representará cada nodo en nuestro árbol.


```python
class Nodo:
    def __init__(self, key):
        self.izq = None
        self.der = None
        self.val = key
```

## Creación de un Árbol

A continuación, crearemos un árbol simple para ilustrar cómo se pueden conectar los nodos.


```python
# Crear el nodo raíz
raiz = Nodo(1)

# Añadir nodos hijos
raiz.izq = Nodo(2)
raiz.der = Nodo(3)

# Añadir nodos hijos al nodo izquierdo del raíz
raiz.izq.izq = Nodo(4)
raiz.izq.der = Nodo(5)

# Añadir nodos hijos al nodo derecho del raíz
raiz.der.izq = Nodo(6)
raiz.der.der = Nodo(7)

# Añadir nodos hijos al nodo hijo derecho del hijo izquierdo del raíz
raiz.izq.der.izq = Nodo(8)
raiz.izq.der.der = Nodo(9)

# Añadir nodos hijos al nodo hijo derecho del hijo derecho del raíz
raiz.der.der.izq = Nodo(10)
raiz.der.der.der = Nodo(11)
```

En el codigo anterior se creo el árbol de ejemplo inicial, en principio cargandolo por _niveles_. Veremos mas adelante otras formas de recorrer un árbol.

## Travesía de un Árbol

Existen diferentes maneras de recorrer un árbol. Aquí mostraremos tres métodos comunes: preorden, inorden y postorden.


### Travesía en Preorden

En la travesía en preorden, visitamos primero la raíz, luego el subárbol izquierdo, y finalmente el subárbol derecho.


```python
def print_preorder(raiz):
    if raiz:
        print(raiz.val, end=" ")
        print_preorder(raiz.izq)
        print_preorder(raiz.der)
        
print_preorder(raiz)  # Salida esperada: 1 2 4 5 8 9 3 6 7 10 11 
```

### Travesía en Inorden

En la travesía en inorden, visitamos primero el subárbol izquierdo, luego la raíz, y finalmente el subárbol derecho.


```python
def print_inorder(raiz):
    if raiz:
        print_inorder(raiz.izq)
        print(raiz.val, end=" ")
        print_inorder(raiz.der)
        
print_inorder(raiz)  # Salida esperada: 4 2 8 5 9 1 6 3 10 7 11
```

### Travesía en Postorden

En la travesía en postorden, visitamos primero el subárbol izquierdo, luego el subárbol derecho, y finalmente la raíz.


```python
def print_postorder(raiz):
    if raiz:
        print_postorder(raiz.izq)
        print_postorder(raiz.der)
        print(raiz.val, end=" ")
        
print_postorder(raiz)  # Salida esperada: 4 8 9 5 2 6 10 11 7 3 1 
```

## Consejos
Puedes crear una clase Nodo para representar los nodos en tu árbol de expresiones. Cada nodo debería tener un valor y dos hijos (izquierdo y derecho).
Puedes crear funciones auxiliares para ayudarte a construir y evaluar el árbol de expresiones.
Recuerda seguir las reglas de precedencia de operadores al construir el árbol.

```python

# ==========================================
# DESAFÍO 76: ÁRBOL BINARIO Y RECORRIDO PREORDER
# ==========================================

class NodoArbol:

    def __init__(self, valor):
        self.valor = valor
        self.izquierdo = None
        self.derecho = None


def preorder(raiz):
    if raiz is not None:
        print(raiz.valor, end=" ")
        preorder(raiz.izquierdo)
        preorder(raiz.derecho)


print("DESAFÍO 76")

raiz = NodoArbol(10)
raiz.izquierdo = NodoArbol(5)
raiz.derecho = NodoArbol(15)

raiz.izquierdo.izquierdo = NodoArbol(3)
raiz.izquierdo.derecho = NodoArbol(7)

raiz.derecho.izquierdo = NodoArbol(12)
raiz.derecho.derecho = NodoArbol(18)

print("Recorrido Preorder:")
preorder(raiz)
print()


# ==========================================
# DESAFÍO 77: RECORRIDO INORDER Y SUMA
# ==========================================

def sumar_inorder(raiz):

    if raiz is None:
        return 0

    suma_izquierda = sumar_inorder(raiz.izquierdo)

    suma_derecha = sumar_inorder(raiz.derecho)

    return suma_izquierda + raiz.valor + suma_derecha


print("\nDESAFÍO 77")

resultado = sumar_inorder(raiz)

print("La suma de todos los nodos es:", resultado)


# ==========================================
# DESAFÍO 78: RECORRIDO POSTORDER Y MAXIMO
# ==========================================

def encontrar_maximo(raiz):

    if raiz is None:
        return None

    maximo_izquierdo = encontrar_maximo(raiz.izquierdo)
    maximo_derecho = encontrar_maximo(raiz.derecho)

    maximo = raiz.valor

    if maximo_izquierdo is not None:
        maximo = max(maximo, maximo_izquierdo)

    if maximo_derecho is not None:
        maximo = max(maximo, maximo_derecho)

    return maximo


print("\nDESAFÍO 78")

print("El valor máximo del árbol es:", encontrar_maximo(raiz))


# ==========================================
# DESAFÍO 79: ÁRBOL DE BÚSQUEDA BINARIA (ABB)
# ==========================================

class NodoABB:

    def __init__(self, valor):
        self.valor = valor
        self.izquierdo = None
        self.derecho = None


def insertar(raiz, valor):

    if raiz is None:
        return NodoABB(valor)

    if valor < raiz.valor:
        raiz.izquierdo = insertar(raiz.izquierdo, valor)

    elif valor > raiz.valor:
        raiz.derecho = insertar(raiz.derecho, valor)

    return raiz


def buscar(raiz, valor):

    if raiz is None:
        return False

    if raiz.valor == valor:
        return True

    if valor < raiz.valor:
        return buscar(raiz.izquierdo, valor)

    return buscar(raiz.derecho, valor)


print("\nDESAFÍO 79")

numeros = [50, 30, 70, 20, 40, 60, 80]

arbol = None

for numero in numeros:
    arbol = insertar(arbol, numero)

numero_buscar = 40

if buscar(arbol, numero_buscar):
    print("El número", numero_buscar, "se encuentra en el árbol")
else:
    print("El número", numero_buscar, "no se encuentra en el árbol")


# ==========================================
# DESAFÍO 80: EVALUACIÓN DE EXPRESIONES
# MATEMÁTICAS CON ÁRBOLES
# ==========================================

import re


class NodoExpresion:

    def __init__(self, valor):
        self.valor = valor
        self.izquierdo = None
        self.derecho = None


def construir_arbol(expresion):

    tokens = re.findall(r'\d+|[()+*/-]', expresion)

    salida = []
    operadores = []

    prioridad = {
        "+": 1,
        "-": 1,
        "*": 2,
        "/": 2
    }

    for token in tokens:

        if token.isdigit():
            salida.append(token)

        elif token == "(":
            operadores.append(token)

        elif token == ")":

            while operadores and operadores[-1] != "(":
                salida.append(operadores.pop())

            operadores.pop()

        else:

            while (operadores and operadores[-1] != "("
                   and prioridad[operadores[-1]] >= prioridad[token]):

                salida.append(operadores.pop())

            operadores.append(token)

    while operadores:
        salida.append(operadores.pop())

    pila = []

    for token in salida:

        nodo = NodoExpresion(token)

        if token in prioridad:
            nodo.derecho = pila.pop()
            nodo.izquierdo = pila.pop()

        pila.append(nodo)

    return pila[0]


def evaluar_arbol(raiz):

    if raiz.izquierdo is None and raiz.derecho is None:
        return int(raiz.valor)

    izquierda = evaluar_arbol(raiz.izquierdo)
    derecha = evaluar_arbol(raiz.derecho)

    if raiz.valor == "+":
        return izquierda + derecha

    elif raiz.valor == "-":
        return izquierda - derecha

    elif raiz.valor == "*":
        return izquierda * derecha

    elif raiz.valor == "/":
        if derecha == 0:
            raise ZeroDivisionError("No se puede dividir por cero")
        return izquierda / derecha


print("\nDESAFÍO 80")

expresion = "5 + 3 * 4"

arbol_expresion = construir_arbol(expresion)

resultado = evaluar_arbol(arbol_expresion)

print("Expresión:", expresion)
print("Resultado:", resultado)
```
